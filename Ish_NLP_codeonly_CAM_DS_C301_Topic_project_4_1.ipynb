```python
# Performing Data preprocessing(Change to lower case, removing stop words, removing numbers and non-alphabets)
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    if pd.isna(text):
        return []
    # Convert to lower case
    text = text.lower()
    # Remove numbers
    text = re.sub(r'\d+', '', text)
    # Tokenize
    tokens = word_tokenize(text)
    # Remove stopwords and non-alphabetic words
    tokens = [word for word in tokens if word.isalpha() and word not in stop_words]
    return tokens
```

```python
# Google dataset
Google_AM["Comment_clean"] = Google_AM["Comment"].apply(preprocess_text)

# Trustpilot dataset
Trustpilot_AM["ReviewContent_clean"] = Trustpilot_AM["Review Content"].apply(preprocess_text)
```

**Conducting initial topic modelling**

Negative reviews for common locations were first extracted and converted into lists to prepare the text for combined analysis and topic modeling across Google and Trustpilot datasets.

```python
#Filtering out reviews that are from location common to both the data sets
common_locations = set(Google_negative["Club's Name"]).intersection(
    set(Trustpilot_negative["Location Name"])
)

Google_negative_common = Google_negative[
    Google_negative["Club's Name"].isin(common_locations)
]
Trustpilot_negative_common = Trustpilot_negative[
    Trustpilot_negative["Location Name"].isin(common_locations)
]
```

```python
print("Google negative reviews (common locations):", Google_negative_common.shape)
print("Trustpilot negative reviews (common locations):", Trustpilot_negative_common.shape)
```

```python
#Converting into list(Google data set)

Google_negative_reviews = Google_negative_common["Comment_clean"].tolist()
```

```python
#Converting into list(Trustpilot data set)

Trustpilot_negative_reviews = Trustpilot_negative_common["ReviewContent_clean"].tolist()
```

```python
#Merging the negative reviews

merged_negative_reviews = Google_negative_reviews + Trustpilot_negative_reviews
```

```python
print("Merged negative reviews:", len(merged_negative_reviews))
```

```python
#BERTopic Model Intialization for Topic Modelling on Preprocessed reviews

topic_model = BERTopic(
    language="english",
    calculate_probabilities=True,
    verbose=True
)
```

```python
# Convert list of lists to list of strings for BERTopic and fitting the model
processed_reviews = [" ".join(review_tokens) for review_tokens in merged_negative_reviews]
topics, probs = topic_model.fit_transform(processed_reviews)
```

```python
topic_info = topic_model.get_topic_info()
```

```python
topic_info.head(10)
```

```python
# Remove outliers (-1) and sort by number of reviews
top_topics = topic_info[topic_info['Topic'] != -1].sort_values(by='Count', ascending=False)

# Take top 2 topics
top_2_topics = top_topics.head(2)
top_2_topics
```

```python
#Top words for the two most frequent topics(BERTopic)

for index, row in top_2_topics.iterrows():
    topic_id = row['Topic']
    top_words = topic_model.get_topic(topic_id)
    # Extract only the words
    top_words_list = [word for word, _ in top_words]
    print(f"Topic {topic_id}: {top_words_list}")
```

**Conducting emotional analysis**

```python
#Initialising BERT-Based Emotion Classification Pipeline

emotion_classifier = pipeline(
    "text-classification",
    model="bhadresh-savani/bert-base-uncased-emotion",
    return_all_scores=True
)
```

```python
emotion_classifier("The staff were rude and the service was terrible.")
```

```python
emotion_classifier("The staff were extremely rude and the service was disappointing.")
```

```python
#Emotion Detection via Chunked Text Processing

def chunk_tokens(tokens, max_tokens=400):
    return [tokens[i:i+max_tokens] for i in range(0, len(tokens), max_tokens)]

def get_top_emotion_no_truncation(text):
    try:
        # Skip empty lists / empty strings
        if text is None or (isinstance(text, list) and len(text) == 0) or (isinstance(text, str) and text.strip() == ""):
            return None

        # Convert token list to string chunks
        if isinstance(text, list):
            chunks = chunk_tokens(text)
            chunk_texts = [" ".join(c) for c in chunks]
        else:
            words = str(text).split()
            if len(words) == 0:
                return None
            chunks = chunk_tokens(words)
            chunk_texts = [" ".join(c) for c in chunks]

        all_preds = []
        for c in chunk_texts:
            preds = emotion_classifier(c)[0]
            all_preds.append(preds)

        # Aggregate across chunks
        labels = all_preds[0]
        avg_scores = {
            lab["label"]: np.mean([p[i]["score"] for p in all_preds])
            for i, lab in enumerate(labels)
        }

        return max(avg_scores, key=avg_scores.get)

    except Exception as e:
        print("Error on first 120 chars:", str(text)[:120])
        print(e)
        return None
```
